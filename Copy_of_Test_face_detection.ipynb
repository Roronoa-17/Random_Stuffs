```python
# Setting Paths
path_video = "/content/test_video.mp4"
path_weight = "/content/yolov8m_200e.pt"
```

```python
# imports
%pip install ultralytics
import ultralytics

ultralytics.checks()
from ultralytics import YOLO
```

```python
# Loading weight for face detection
model = YOLO(path_weight)
```

### Tracking and Saving output video (https://docs.ultralytics.com/modes/track/#tracking)

```python
# Localize and tracking face
results = model.track(source=path_video,  conf=0.5, iou=0.5, show=True, save=True)
```

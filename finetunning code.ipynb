```python
from ultralytics import YOLO
model = YOLO('yolov8n.pt')
model.train(data='/content/drive/MyDrive/YOLO-HiVis-Data/data.yaml', epochs=30, imgsz=640, batch=16, name='yolov8_hivis')
```

```python
import cv2
from matplotlib import pyplot as plt

# Load the model (specify the correct path if necessary)
model = YOLO('/content/drive/MyDrive/runs/detect/yolov8_hivis2/weights/best.pt')

# Predict on the validation dataset
val_images_path = '/content/drive/MyDrive/YOLO-HiVis-Data/temp'
predictions = model(val_images_path, save=True)
predicted_images_path = 'runs/detect/predict2'

# List all images in the prediction directory
import os
predicted_images = [os.path.join(predicted_images_path, img) for img in os.listdir(predicted_images_path) if img.endswith(('.png', '.jpg', '.jpeg'))]

# Display a few sample images with predictions
for img_path in predicted_images[:5]:  # Display first 5 images
    img = cv2.imread(img_path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    plt.imshow(img)
    plt.axis('off')
    plt.show()
```
